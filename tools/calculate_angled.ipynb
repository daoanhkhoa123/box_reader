```python
import math
from dataclasses import dataclass

@dataclass
class Params:
    focal_length: float
    box_width: float
    dist_to_mid_point: float
    angle: float

def __calculate_other_side_squared(params: Params, with_angle: float):
    box_width_half = params.box_width / 2 
    return params.dist_to_mid_point ** 2 + box_width_half ** 2 - 2 * params.dist_to_mid_point * box_width_half * math.cos(with_angle)

def _calculate_two_angles(params: Params):
    box_width_half = params.box_width / 2 

    ao2 = __calculate_other_side_squared(params, math.radians(90 - params.angle))
    cos_aoz = (ao2 + params.dist_to_mid_point ** 2 - box_width_half ** 2) / (2 * math.sqrt(ao2) * params.dist_to_mid_point)

    bo2 = __calculate_other_side_squared(params, math.radians(90 + params.angle))
    cos_boz = (bo2 + params.dist_to_mid_point ** 2 - box_width_half ** 2) / (2 * math.sqrt(bo2) * params.dist_to_mid_point)

    return math.acos(cos_aoz), math.acos(cos_boz)

def calculate_image_width(params: Params):
    aoz, boz = _calculate_two_angles(params)
    return (math.tan(aoz) + math.tan(boz)) * params.focal_length, math.degrees(aoz), math.degrees(boz)

```

```python
p1 = Params(focal_length=1, box_width=0.5, dist_to_mid_point= 1, angle=47)
w, a1, a2 = calculate_image_width(p1)

for k in range(1, 60):
    p = Params(focal_length=1 * k, box_width=0.5, dist_to_mid_point= 1, angle=47)
    w2, a1, a2 = calculate_image_width(p)
    print(w2 / p.focal_length)
```
